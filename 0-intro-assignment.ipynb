```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
sns.set_theme(style="darkgrid")
```

```python
data = {
    "Square_Feet": np.random.randint(500, 3500, size=10),
    "Location": ["Rural", "Urban", "Suburban", "Rural", "Rural", "Urban", "Suburban", "Urban", "Rural", "Suburban"],
    "Price": np.random.uniform(low=1000, high=5000, size=10)
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.describe()
```

```python
avg_price_per_location = df.groupby('Location')['Price'].mean()
print(avg_price_per_location)
```
